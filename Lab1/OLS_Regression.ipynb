```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. ข้อมูลจากโจทย์ (X: Budget, Y: Sales)
X = np.array([3.0, 5.0, 2.0, 7.0, 8.0, 1.0, 4.0, 6.0, 9.0, 10.0])
Y = np.array([6.0, 9.0, 4.0, 10.0, 12.0, 3.0, 7.0, 8.0, 13.0, 15.0])

# 2. คำนวณหา Slope (beta_1) และ Intercept (beta_0) ด้วยวิธี OLS
n = len(X)
X_mean = np.mean(X)
Y_mean = np.mean(Y)

beta_1 = np.sum((X - X_mean) * (Y - Y_mean)) / np.sum((X - X_mean)**2)
beta_0 = Y_mean - (beta_1 * X_mean)

# 3. สร้างฟังก์ชันทำนายผล
def predict(x):
    return beta_0 + (beta_1 * x)

# 4. คำนวณ R-squared (R2)
Y_pred = predict(X)
SS_res = np.sum((Y - Y_pred)**2)
SS_tot = np.sum((Y - Y_mean)**2)
r_squared = 1 - (SS_res / SS_tot)

# 5. ทำนายยอดขายเมื่อ งบโฆษณา = $12,000 (X = 12)
X_new = 12.0
Y_new_pred = predict(X_new)

# แสดงผลลัพธ์การคำนวณ
print(f"Slope (beta_1)     : {beta_1:.4f}")
print(f"Intercept (beta_0) : {beta_0:.4f}")
print(f"Equation           : Y = {beta_0:.4f} + {beta_1:.4f}*X")
print(f"Predicted Sales for $12k budget (X=12): {Y_new_pred:.4f} (thousand units)")
print(f"R-squared Value    : {r_squared:.4f}")

# 6. วาดกราฟแสดงผล
plt.figure(figsize=(8, 5))
plt.scatter(X, Y, color='blue', label='Actual Data Points')
plt.plot(X, Y_pred, color='red', label=f'Fitted Line: Y = {beta_0:.2f} + {beta_1:.2f}X')
plt.scatter([X_new], [Y_new_pred], color='green', s=100, zorder=5, label=f'Prediction at X=12 (Y={Y_new_pred:.2f})')
plt.xlabel('Advertising Budget (in $1,000s)')
plt.ylabel('Sales (in 1,000s units)')
plt.title('Simple Linear Regression (OLS) - Advertising vs Sales')
plt.legend()
plt.grid(True)
plt.show()
```

## Task 1: Implement OLS Regression
1. **Regression Coefficients:**
   - **Slope ($\beta_1$):** `1.2667`
   - **Intercept ($\beta_0$):** `1.7333`
2. **Equation of the fitted line:**
   $$\hat{Y} = 1.7333 + 1.2667X$$

---

## Task 2: Interpret the Results
1. **Meaning of Slope and Intercept:**
   - **Slope ($\beta_1 = 1.2667$):** หมายความว่า ทุกๆ การเพิ่มขึ้นของงบประมาณโฆษณา $1,000 ($X$ เพิ่มขึ้น 1 หน่วย) คาดว่าจะทำให้ยอดขายเพิ่มขึ้นประมาณ **1,267 ชิ้น** ($Y$ เพิ่มขึ้นประมาณ 1.2667 หน่วย)
   - **Intercept ($\beta_0 = 1.7333$):** หมายความว่า หากไม่มีการใช้งบประมาณโฆษณาเลย ($X = 0$) ยอดขายพื้นฐานที่คาดว่าจะเกิดขึ้นคือประมาณ **1,733 ชิ้น** ($1.7333 \times 1,000$)
2. **Sales Prediction for $12,000 Advertising Budget ($X = 12$):**
   - แทนค่า $X = 12$ ลงในสมการ:
     $$\hat{Y} = 1.7333 + 1.2667(12) = 16.9333$$
   - **คำตอบ:** คาดการณ์ว่าจะมียอดขายประมาณ **16,933 ชิ้น** ($16.9333 \times 1,000$)
